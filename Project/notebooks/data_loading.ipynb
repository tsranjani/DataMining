```python
import os
import pandas as pd
pd.set_option('display.max_colwidth', 500)

DATA_DIR = os.path.join("..", "data")
```

```python
def print_info(df: pd.DataFrame):
    print('# of rows:   ', df.shape[0])
    print('# of columns:', df.shape[1])

    print('\nColumn data types:')
    for items in df.dtypes.value_counts().items():
        print(f'  # of {items[0]} columns: {items[1]}')
        
    print('\nUnique values per column (only first 5 columns are shown):')
    for col in df.columns[:5]:  # limit to first 5 columns
        print(f'  {col}: {df[col].nunique()}')
    print('  ...')
```

## World Bank Data

```python
# GDP per capita
# main file + 2 metadata files (not too important for us)
gdp_per_capita_df = pd.read_csv(f'{DATA_DIR}/gdp_per_capita/gdp_per_capita_data.csv', skiprows=4)
print_info(gdp_per_capita_df)
```
